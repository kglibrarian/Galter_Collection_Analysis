```python
## Step 50: Function to check if data is present in the column
def check_ORCID_at_NU_present(value):
    if pd.isna(value) or value == '':
        return 'N'
    else:
        return 'Y'

# Apply the function to create a new column
elements_26_df['Completed Connection between ORCID and NetID?'] = elements_26_df['ORCID'].apply(lambda x: check_ORCID_at_NU_present(x))
```

```python
# Step 51: Function to check if ORCID data is present in the column
def check_name_match_at_ORCID_present(value):
    if pd.isna(value) or value == '':
        return 'N'
    else:
        return 'Y'

# Apply the function to create a new column
elements_26_df['Possible name match at ORCID.org?'] = elements_26_df['Name Match at ORCID.org?'].apply(lambda x: check_name_match_at_ORCID_present(x))
```

## Custom Processing for this report process

This section uses functions written specifically for this report to:
- Drop duplicates for multi-affiliated individuals
- Merge Alma and Elements datasets
- Export to Excel


```python
# Step 2: Merge the dataframes on 'NetID' and 'Primary Identifier'
merged_df = pd.merge(elements_29_df, df2, left_on='NetID', right_on='Primary Identifier', how='inner')
merged_df.head()
```

```python
# Step 3: Identify unmatched rows from df2 (those that didn’t match with df1)
unmatched_df2 = df2[~df2['Primary Identifier'].isin(elements_29_df['NetID'])]
unmatched_df2.head()
```

```python
# Step 4: Export both to an Excel file with two sheets
output_path = r"output\Alma_Galter_Users_Match\2025_06 June\2025_06-15_Alma_Galter_Users_Match_with_Elements.xlsx"
with pd.ExcelWriter(output_path, engine='xlsxwriter') as writer:
    merged_df.to_excel(writer, sheet_name='Matched Records', index=False)
    unmatched_df2.to_excel(writer, sheet_name='Unmatched Records', index=False)

print(f"Excel workbook saved to: {output_path}")
```
